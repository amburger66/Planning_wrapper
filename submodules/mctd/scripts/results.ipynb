```python
import os
import numpy as np

dir = "../inference_block/ckpt300k_se"
scales = [1, 2, 10, 20]

print("--- Guided Diffuser ---")

for scale in scales:
    successes = 0
    total = 0
    for f in os.listdir(os.path.join(dir, "guided", f"scale_{scale}")):
        if f.endswith(".npz"):
            data = np.load(os.path.join(dir, "guided", f"scale_{scale}", f))
            # print(data["solved"])
            successes += data["solved"]
            total += 1
    print(f"Scale {scale}: {successes}/{total}")


print("--- MCTD ---")
successes = 0
total = 0
for f in os.listdir(os.path.join(dir, "mctd")):
    if f.endswith(".npz"):
        data = np.load(os.path.join(dir, "mctd", f))
        # print(data["solved"])
        successes += data["solved"]
        total += 1
print(f"MCTD: {successes}/{total}")
```
